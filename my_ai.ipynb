```python
from pypdf import PdfReader

def read_pdf(pdf_path):
    reader = PdfReader(pdf_path)

    pages = []

    for page_number, page in enumerate(reader.pages, start=1):
        page_text = page.extract_text()

        if page_text:
            pages.append({
                "page": page_number,
                "text": page_text
            })

    return pages
```

```python
import os

def load_all_pdfs(folder_path):
    all_pages = []

    for filename in os.listdir(folder_path):
        if filename.lower().endswith(".pdf"):
            pdf_path = os.path.join(folder_path, filename)

            pages = read_pdf(pdf_path)

            for page in pages:
                all_pages.append({
                    "filename": filename,
                    "page": page["page"],
                    "text": page["text"]
                })

    return all_pages
```

```python
def split_text(text, chunk_size=1000):
    # Remove excessive whitespace
    text = "\n".join(
        line.strip()
        for line in text.splitlines()
        if line.strip()
    )

    paragraphs = text.split("\n\n")

    chunks = []
    current_chunk = ""

    for paragraph in paragraphs:
        if len(current_chunk) + len(paragraph) <= chunk_size:
            current_chunk += paragraph + "\n\n"
        else:
            if current_chunk.strip():
                chunks.append(current_chunk.strip())

            current_chunk = paragraph + "\n\n"

    if current_chunk.strip():
        chunks.append(current_chunk.strip())

    # Remove very small chunks
    chunks = [
        chunk for chunk in chunks
        if len(chunk) >= 100
    ]

    return chunks
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def search_pdf(query, all_chunks, chunk_embeddings, top_k=8):

    query_embedding = embedding_model.encode([query])

    similarities = cosine_similarity(
        query_embedding,
        chunk_embeddings
    )[0]

    query_words = set(query.lower().split())

    results = []

    for index, chunk in enumerate(all_chunks):

        text_lower = chunk["text"].lower()

        # Count important query words found in the chunk
        keyword_matches = sum(
            1 for word in query_words
            if len(word) > 3 and word in text_lower
        )

        # Combine semantic similarity and keyword matching
        final_score = similarities[index] + (keyword_matches * 0.05)

        results.append({
            "text": chunk["text"],
            "source": chunk["source"],
            "page": chunk["page"],
            "score": final_score
        })

    # Sort by final score
    results.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return results[:top_k]
```

```python
def answer_from_pdf(query):
    results = search_pdf(
        query,
        all_chunks,
        chunk_embeddings,
        top_k=5
    )

    context = "\n\n".join(
        f"Source: {result['source']}, Page: {result['page']}\n{result['text']}"
        for result in results
    )

    prompt = f"""
You are a helpful AI assistant with access to multiple PDFs.

Use the PDF context below to answer the user's question.

PDF CONTEXT:
{context}

USER QUESTION:
{query}

Give a clear answer based on the PDF context.
"""

    response = ollama.chat(
        model="gemma3",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        stream=True
    )

    full_response = ""

    for chunk in response:
        text = chunk["message"]["content"]
        print(text, end="", flush=True)
        full_response += text

    sources = list(dict.fromkeys(
        (result["source"], result["page"])
        for result in results
    ))

    print("\n\nSources:")
    for source, page in sources:
        print(f"📄 {source} — Page {page}")

    return full_response
```

```python
import gradio as gr


def clear_chat():
    global conversation

    conversation = [
        {
            "role": "system",
            "content": "You are a helpful AI assistant. Give clear, accurate and beginner-friendly responses."
        }
    ]

    save_conversation(conversation)

    return []


def chat_with_ai(message, history):

    response = answer_question(message, conversation)

    # Get the relevant PDF sources
    results = search_pdf(
        message,
        all_chunks,
        chunk_embeddings,
        top_k=5
    )

    sources = list(dict.fromkeys(
        (result["source"], result["page"])
        for result in results
    ))

    source_text = "\n\n📚 Sources:\n"

    for source, page in sources:
        source_text += f"📄 {source} — Page {page}\n"

    response_with_sources = response + source_text

    add_message(conversation, "assistant", response)
    save_conversation(conversation)

    history = history or []

    history.append({
        "role": "user",
        "content": message
    })

    history.append({
        "role": "assistant",
        "content": response_with_sources
    })

    return history


with gr.Blocks() as demo:

    gr.Markdown("# 🤖 My AI Assistant")
    gr.Markdown(
        "Local AI assistant with PDF knowledge and persistent memory."
    )

    pdf_upload = gr.File(
        label="📄 Upload a PDF",
        file_types=[".pdf"]
    )

    pdf_status = gr.Textbox(
        label="PDF Status",
        interactive=False
    )

    pdf_upload.change(
        process_pdf,
        inputs=pdf_upload,
        outputs=pdf_status
    )

    chatbot = gr.Chatbot(
    label="AI Assistant"
    )

    message = gr.Textbox(
        placeholder="Ask me anything...",
        label="Your message"
    )

    send = gr.Button("Send")
    clear = gr.Button("🧹 Clear Chat")

    send.click(
        chat_with_ai,
        inputs=[message, chatbot],
        outputs=chatbot
    )

    message.submit(
        chat_with_ai,
        inputs=[message, chatbot],
        outputs=chatbot
    )

    clear.click(
        clear_chat,
        outputs=chatbot
    )


demo.launch()
```

```python
def process_pdf(file):
    global documents, all_chunks, chunk_embeddings

    try:
        if isinstance(file, str):
            pdf_path = file
        elif hasattr(file, "name"):
            pdf_path = file.name
        elif isinstance(file, dict):
            pdf_path = file["path"]
        else:
            raise ValueError(f"Unsupported file type: {type(file)}")

        # Read PDF pages
        pages = read_pdf(pdf_path)

        # Add filename to every page
        filename = os.path.basename(pdf_path)

        documents = []

        for page in pages:
            documents.append({
                "filename": filename,
                "page": page["page"],
                "text": page["text"]
            })

        # Create chunks
        all_chunks = create_document_chunks(documents)

        # Create embeddings
        chunk_texts = [chunk["text"] for chunk in all_chunks]

        chunk_embeddings = embedding_model.encode(
            chunk_texts,
            show_progress_bar=True
        )

        return f"PDF loaded successfully! {len(documents)} pages, {len(all_chunks)} chunks."

    except Exception as e:
        print("ERROR:", repr(e))
        return f"ERROR: {repr(e)}"
```
